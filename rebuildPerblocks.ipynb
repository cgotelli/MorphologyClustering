# Matrix rebuilder
Rebuild final matrix from chunks obtained with MHD-EGU-perblocks-every02

```python
import os
import numpy as np
import re

# Variables iniciales
matrix_name = "DeepestRemovedBlobs-every02-EGU"
folder_path = r"X:/PhD_Gotelli/Exp20/EGU24Code/DeepestRemovedBlobs-every02/imgs_filtered"
# Ruta de la carpeta donde están los chunks
chunks_dir = os.path.join(folder_path, "../blocks")

# Obtener la lista de archivos de chunks
chunk_files = sorted(
    [
        f
        for f in os.listdir(chunks_dir)
        if f.startswith(f"{matrix_name}_chunk_") and f.endswith(".npy")
    ]
)

# Extraer información de cada chunk (start_row, end_row)
chunk_info = []
pattern = re.compile(rf"{re.escape(matrix_name)}_chunk_(\d+)_(\d+)\.npy")

for file in chunk_files:
    match = pattern.match(file)
    if match:
        start_row = int(match.group(1))
        end_row = int(match.group(2))
        chunk_info.append((start_row, end_row, file))

# Determinar N, el número total de filas/columnas
N = 0
for start_row, end_row, _ in chunk_info:
    if end_row + 1 > N:
        N = end_row + 1

# Inicializar la matriz completa en memoria
distance_matrix = np.zeros((N, N), dtype=np.float64)

# Procesar cada chunk
for start_row, end_row, file in chunk_info:
    print(f"Procesando chunk {file} (filas {start_row} a {end_row})")
    chunk_path = os.path.join(chunks_dir, file)
    # Cargar el chunk
    chunk = np.load(chunk_path)
    
    # Asegurarse de que las dimensiones del chunk sean correctas
    num_rows_in_chunk = end_row - start_row + 1
    assert chunk.shape[0] == num_rows_in_chunk, "El número de filas en el chunk no coincide."
    
    # Insertar el chunk en la matriz completa
    distance_matrix[start_row:end_row+1, :] = chunk

# Completar la matriz simétricamente
for i in range(N):
    print(f"Completando simetría en fila {i+1}/{N}")
    distance_matrix[i+1:N, i] = distance_matrix[i, i+1:N]

# Guardar la matriz completa
distance_matrix_file = os.path.join(chunks_dir, f"{matrix_name}_full_distance_matrix.npy")
np.save(distance_matrix_file, distance_matrix)

print("Reconstrucción y completado de la matriz completa finalizados.")
```
